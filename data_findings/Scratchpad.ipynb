```python
trips = pd.read_csv("./data/trips_small.csv", sep = ',')
cities = pd.read_csv("./data/cities.csv", sep = ',')
bike_types = pd.read_csv("./data/bike_types.csv", sep = ',')
bikes = pd.read_csv("./data/bikes.csv", sep = ',')
stations = pd.read_csv("./data/stations.csv", sep = ',')
station_status = pd.read_csv("./data/stations_status.csv", sep = ',')
```

```python
datasets = {
    "trips": trips,
    "cities": cities,
    "bike_types": bike_types,
    "bikes": bikes,
    "stations": stations,
    "station_status": station_status,
}

def describe_df(name, df, n=5):
    print(f"\n{'=' * 80}\n{name}  —  shape: {df.shape}\n{'=' * 80}")

    print("\n-- head --")
    print(df.head(n).to_string())

    print("\n-- column overview --")
    overview = pd.DataFrame({
        "dtype": df.dtypes.astype(str),
        "na_count": df.isna().sum(),
        "na_pct": (df.isna().mean() * 100).round(2),
        "n_unique": df.nunique(dropna=True),
    })
    print(overview.to_string())

    print("\n-- summary (numeric) --")
    num = df.select_dtypes(include="number")
    print(num.describe().T.to_string() if not num.empty else "(no numeric columns)")

    print("\n-- summary (non-numeric) --")
    obj = df.select_dtypes(exclude="number")
    print(obj.describe().T.to_string() if not obj.empty else "(no non-numeric columns)")

for name, df in datasets.items():
    describe_df(name, df)
```

```python
other_cols = bike_types.columns.drop("id")
bike_types["type_info_missing"] = bike_types[other_cols].isna().all(axis=1)

# Copy propulsion_type, and assume "human" for the empty types
bike_types["propulsion_filled"] = bike_types["propulsion_type"]
bike_types.loc[bike_types["type_info_missing"], "propulsion_filled"] = "human"

# Text columns: fill with "unknown"
for col in ["vehicle_image", "name", "description", "form_factor", "propulsion_type"]:
    bike_types[col] = bike_types[col].fillna("unknown")

# Rider capacity: 1 when unknown
bike_types["rider_capacity"] = bike_types["rider_capacity"].fillna(1).astype(int)

# Range and battery: 0 for all human bikes (original and backfilled)
is_human = bike_types["propulsion_filled"] == "human"
for col in ["max_range", "battery_capacity"]:
    bike_types.loc[is_human, col] = 0

# id 238, 240: DOLLY BIKES "e-Dolly" MaxDrive -> "battery 600 Wh 43 V under bench"
bike_types.loc[bike_types["id"].isin([238, 240]), "battery_capacity"] = 600

# id 179, 181: same eHarry description as 121 and 224, which have 630 Wh
bike_types.loc[bike_types["id"].isin([179, 181]), "battery_capacity"] = 630
```

```python
# Rename the id columns so they match the foreign keys and don't clash
bikes = bikes.rename(columns={"id": "bike_id"})
bike_types = bike_types.rename(columns={"id": "bike_type_id", "name": "bike_type_name"})

# Drop columns that are not useful per trip (long text, URLs)
bike_types = bike_types.drop(columns=["vehicle_image", "description"])

trips_full = trips.merge(bikes, on="bike_id", how="left", validate="many_to_one")
trips_full = trips_full.merge(bike_types, on="bike_type_id", how="left", validate="many_to_one")
```

```python
trips_full["has_battery"] = trips_full["battery_start"].notna() | trips_full["battery_end"].notna()

human = trips_full[trips_full["propulsion_filled"] == "human"]

per_bike = human.groupby("bike_id").agg(
    trips=("has_battery", "size"),
    battery_trips=("has_battery", "sum"),
    bike_type_id=("bike_type_id", "first"),
)
per_bike["battery_share"] = per_bike["battery_trips"] / per_bike["trips"]
with_battery = per_bike[per_bike["battery_trips"] > 0]
```

```python
battery_bike_ids = with_battery.index

cols = [
    "bike_id", "bike_type_id",
    "battery_start", "battery_end", "has_battery",
]

human_battery_trips = trips_full.loc[trips_full["bike_id"].isin(battery_bike_ids), cols].copy()

# Readable time and battery used per trip
human_battery_trips["battery_used"] = human_battery_trips["battery_start"] - human_battery_trips["battery_end"]

human_battery_trips = human_battery_trips.sort_values(["bike_id"])
human_battery_trips[human_battery_trips['battery_used'] > 0]
# Only bike with ID 31742 has trips with actual battery usage
```
